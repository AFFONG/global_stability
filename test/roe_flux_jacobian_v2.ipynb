```python
# finding dF/dUj , dF/dUi
"""
3-D Convective Flux Jacobian (Equation A.47 / A.48)
Symbolic implementation using SymPy.
 
References
----------
Formulas from Appendix A.7 Blazek's book, equations (A.47) and (A.48).
"""
 
import sympy as sp
from sympy import init_printing
init_printing()
 
# ── Symbols ──────────────────────────────────────────────────────────────────
nx, ny, nz = sp.symbols('n_x n_y n_z', real=True)   # face-normal components
u,  v,  w  = sp.symbols('u v w',       real=True)   # velocity components
E           = sp.Symbol('E',            positive=True)  # total energy
gamma       = sp.Symbol('gamma',        positive=True)  # ratio of specific heats
 
# ── Abbreviations (A.48) ─────────────────────────────────────────────────────
phi = sp.Rational(1, 2) * (gamma - 1) * (u**2 + v**2 + w**2)
V   = nx*u + ny*v + nz*w                          # normal velocity
a1  = gamma*E - phi
a2  = gamma - 1
a3  = gamma - 2
 
abbreviations = {
    'phi': phi,
    'V'  : V,
    'a1' : a1,
    'a2' : a2,
    'a3' : a3,
}
 
# ── dFdU (A.47) ─────────────────────────────────────
# Rows are listed top-to-bottom as they appear in the paper.
# The matrix is 5×5 (continuity + 3 momenta + energy).
 
dFdU = sp.Matrix([
    # Row 1
    [ -V,
       nx,
       ny,
       nz,
       0  ],
 
    # Row 2
    [ nx*phi - u*V,
       V - a3*nx*u,
       ny*u - a2*nx*v,
       nz*u - a2*nx*w,        # paper uses n_z u - a_2 n_x w
       a2*nx  ],
 
    # Row 3
    [ ny*phi - v*V,
       nx*v - a2*ny*u,
       V - a3*ny*v,
       nz*v - a2*ny*w,
       a2*ny  ],
 
    # Row 4
    [ nz*phi - w*V,
       nx*w - a2*nz*u,
       ny*w - a2*nz*v,
       V - a3*nz*w,
       a2*nz  ],
 
    # Row 5
    [ V*(phi - a1),
       nx*a1 - a2*u*V,
       ny*a1 - a2*v*V,
       nz*a1 - a2*w*V,
       gamma*V ],
])
 
```
